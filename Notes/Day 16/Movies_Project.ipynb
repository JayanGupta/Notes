# Q3. How do we clean the `duration` column? (e.g., `'175 min'` → `175`)


```python
df["duration"] = df["duration"].str.replace('min' , '')
df["duration"] = df["duration"].str.replace(',' , '')
```

```python
df["duration"] = df["duration"].astype(int)
```

```python
df.info()
```

# Q4. How do we clean the `gross` column and `votes` column?


```python
df["votes"] = df["votes"].str.replace(",","")
```

```python
df["votes"] = df["votes"].astype(int)
```

```python
df.info()
```

```python
df['gross_income'] = df['gross_income'].str.replace(",","")
df['gross_income'] = df['gross_income'].str.replace("$","")
df['gross_income'] = df['gross_income'].str.replace("M","")
```

```python
df['gross_income'] = df['gross_income'].astype(float)
```

```python
df.info()
```

# Q1. Plot the average movie rating over the years (Line Chart)


```python
a  = df.groupby("year")["rating"].mean()
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize = (12,6))

plt.plot(a.index , a.values , color = "teal" , marker = "o" , markersize = 5 , linestyle = "-.")

plt.xlabel("Year")
plt.ylabel("Average Rating")

plt.show()
```

# Q2. Plot the number of movies released per year (Line Chart)


```python
mpy = df["year"].value_counts().sort_index()
```

```python
plt.figure(figsize = (12,6))

plt.plot(mpy.index , mpy.values , color = "red" , linestyle = "dashed" , marker = "o" , markersize = 4)
plt.xlabel("Year")
plt.ylabel("Number of Movies")
plt.title("Number of movies released per year")

plt.grid(alpha = 0.25)

plt.show()
```

# Q5. Which are the top 10 most frequent actors? ( Bar Chart)


```python
actors = df["stars_name"].str.split(",").explode().str.strip()

top10 = actors.value_counts().head(10)
```

```python
plt.figure(figsize = (12 ,6))

plt.bar(top10.index , top10.values , width = 0.7 , color = ["red" , "yellow" , "green" , 'cyan' , 'blue' , 'purple'])

plt.xlabel("Actors")
plt.ylabel('Number of Movies')

plt.title("Top 10 most frequent actors")

plt.xticks(rotation = 30)

plt.show()
```

```python
plt.figure(figsize = (12 ,6))

plt.barh(top10.index , top10.values , height = 0.5, color = ["red" , "yellow" , "green" , 'cyan' , 'blue' , 'purple'])

plt.xlabel("Actors")
plt.ylabel('Number of Movies')

plt.title("Top 10 most frequent actors")

plt.show()
```
