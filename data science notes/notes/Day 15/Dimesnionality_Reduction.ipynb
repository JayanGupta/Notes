```python
import matplotlib.pyplot as plt

from sklearn.datasets import make_moons
```

```python
X , y = make_moons(n_samples= 300 , noise = 0.08, random_state = 42)
```

```python
plt.figure(figsize = (10,6))

plt.scatter(X[:,0] ,X[:,1])

plt.show()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.cluster import DBSCAN , KMeans
```

```python
model = DBSCAN( eps = 0.3 , min_samples = 7 )
```

```python
labels = model.fit_predict(X_scaled)
```

```python
print(labels)
```

```python
len(set(labels))
```

```python
plt.figure(figsize = (10,6))

plt.scatter(X[:,0] ,X[:,1], c = labels , cmap = "rainbow")

plt.show()
```

# Independent Componenet Analysis (ICA)

```python
import matplotlib.pyplot as plt
from sklearn.decomposition import FastICA
```

```python
import numpy as np
```

```python
np.random.seed(42)

samples = 200

time = np.linspace(0,8, samples)

s1 = np.sin(2*time)
s2 = np.sign(np.sin(3 * time))
s3 = np.random.laplace(size = samples)

S = np.c_[s1,s2,s3]
S = S + 0.2 * np.random.normal(size = S.shape)
```

```python
A = np.array([[1,1,1] , [0.5,2,1] , [1.5 ,1 ,2]])
```

```python
X = S.dot(A.T)
```

```python
plt.figure(figsize = (12, 6))

plt.title('Original Source')
plt.plot(S)
plt.show()
```

```python
plt.figure(figsize = (12, 6))


plt.title("Observed Signal")
plt.plot(X)
plt.show()
```

```python
model = FastICA(n_components= 3)

SS = model.fit_transform(X)
```

```python
plt.figure(figsize = (12, 6))


plt.title("FastICA Signal")
plt.plot(SS)
plt.show()
```
