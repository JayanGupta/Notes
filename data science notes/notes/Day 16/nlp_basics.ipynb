# Basic NLP with the NLTK Library
Welcome! In this notebook, we'll learn about **Natural Language Processing (NLP)** using Python's famous **NLTK (Natural Language Toolkit)** library. NLP is how we teach computers to understand human languages like English.

**Step 1: Downloading the "Brains" of NLTK**
Before we can use NLTK, we need to download some of its specific toolsets (or data packs). Think of NLTK as a library building, and these downloads as the specific books we are taking off the shelf to read:
- **`punkt` and `punkt_tab`**: These are used by NLTK to understand where words and sentences begin and end. They act like a grammar rulebook for punctuation.
- **`stopwords`**: This is a list of very common English words (like "the", "is", "a") that we often want to ignore when analyzing text.

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

**Question 1: What is Word Tokenization?**
Tokenization is the process of breaking a large piece of text into smaller pieces (called "tokens"). 
How can we split a sentence into individual words?

```python
from nltk.tokenize import word_tokenize

my_sentence = "Natural Language Processing is amazing! Computers are learning to read."
words = word_tokenize(my_sentence)

print("Our original sentence:")
print(my_sentence)
print("\nOur list of word tokens:")
print(words)
```

**Question 2: What is Sentence Tokenization?**
Sometimes we don't want to split by words, but by sentences. 
How can we split a whole paragraph into individual sentences?

```python
from nltk.tokenize import sent_tokenize

paragraph = "Hello everyone! Welcome to our first NLP class. It is a fascinating subject. Are you ready to learn?"
sentences = sent_tokenize(paragraph)

print("Our original paragraph:")
print(paragraph)
print("\nOur list of sentences:")
print(sentences)
```

**Question 3: How do we find the most common word in a text?**
If we have a long string of words, we might want to know which word appears the most. We can use a Frequency Distribution for this.

```python
from nltk.probability import FreqDist

# Let's imagine this is a review for a fruit stand
text = "apple orange banana apple grape apple banana orange apple"
words = word_tokenize(text)

# Create a frequency distribution (it counts how many times each word appears)
fdist = FreqDist(words)

# Find the single most common word
print("The most popular word is:")
print(fdist.most_common(1))
```

**Question 4: What are Stopwords?**
Stopwords are common words that don't add much deep meaning to a sentence (like "the", "is", "in"). 
How can we get a list of them in English so we know what to filter out later?

```python
from nltk.corpus import stopwords

# Get the list of English stopwords
stop_words = stopwords.words('english')

# Print just the first 10 stopwords to see what they look like
print("Here are the first 10 stopwords in the English language:")
print(stop_words[:10])
```
