# Pre-Processing


1. All the text ---> lowercase
2. Tokenise the sentence ---> words
3. Remove Stopwords (the , is , ...)
4. Removing Punctutaion (! , .)
5. Join everything which is remaining.

```python
stop_words = set(stopwords.words("english"))
```

```python
# Function

def preprocess(text):

    text = text.lower()

    words = word_tokenize(text)

    words = [ word
            for word in words
            if word.isalpha() and word not in stop_words ]

    return " ".join(words)
```

```python
sample = df["text"][0]

print(sample)
```

```python
pro = preprocess(sample)

print(pro)
```

```python
# Training Data

df["clean_text"] = df['text'].apply(preprocess)

# Testing Data

df2["clean_text"] = df2['text'].apply(preprocess)
```

```python
df.head()
```

```python
X_train = df['clean_text']

X_test =  df2['clean_text']

y_train = df['category']

y_test = df2['category']
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vec = TfidfVectorizer( max_features = 10000)
```

```python
X_train_tf = vec.fit_transform(X_train)

X_test_tf = vec.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter= 1000)

model.fit(X_train_tf , y_train)
```

```python
y_pred = model.predict(X_test_tf)
```

```python
y_pred[:10]
```

```python
from sklearn.metrics import accuracy_score , classification_report

acc = accuracy_score(y_test , y_pred)

print(acc * 100)
```
