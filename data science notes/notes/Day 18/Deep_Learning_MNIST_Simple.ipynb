# Deep Learning with TensorFlow — Handwritten Digit Recognition

In this notebook, we will build a simple neural network that learns to recognize handwritten digits using the MNIST dataset.

### What we will do
1. Load the MNIST dataset
2. Look at the images and labels
3. Understand pixel values
4. Normalize the data
5. Build a simple neural network
6. Train the model
7. Evaluate the model
8. Make a prediction


## Step 1 — Import the Libraries

```python
import tensorflow as tf
import matplotlib.pyplot as plt

print("TensorFlow version:", tf.__version__)
```

## Step 2 — Load the MNIST Dataset

```python
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.mnist.load_data()

print("Training images:", X_train.shape)
print("Training labels:", y_train.shape)
print("Testing images:", X_test.shape)
print("Testing labels:", y_test.shape)
```

## Step 3 — Look at the First Image

Before building a model, let's first see what our data actually looks like.

```python
plt.imshow(X_train[0], cmap="gray")
plt.title("First Training Image")
plt.axis("off")
plt.show()

print("Label:", y_train[0])
```

## Step 4 — Look at Another Image

```python
plt.imshow(X_train[1], cmap="gray")
plt.title("Second Training Image")
plt.axis("off")
plt.show()

print("Label:", y_train[1])
```

## Step 5 — Understand the Pixel Values

Each image is a 28 × 28 grid of numbers.

The pixel values are between 0 and 255:
- 0 means black
- 255 means white
- Values between them represent different shades of gray

```python
print(X_train[0])
```

```python
print("Minimum pixel value:", X_train[0].min())
print("Maximum pixel value:", X_train[0].max())
```

## Step 6 — Normalize the Data

Neural networks work better when the input values are small.

We divide every pixel by 255 so that the values become between 0 and 1.

```python
X_train = X_train / 255.0
X_test = X_test / 255.0

print("Minimum value:", X_train.min())
print("Maximum value:", X_train.max())
```

## Step 7 — Build the Neural Network

Our model will have:
- A Flatten layer to convert the 28 × 28 image into numbers
- A hidden layer with 128 neurons
- An output layer with 10 neurons, one for each digit (0–9)

```python
model = tf.keras.Sequential([
    tf.keras.layers.Flatten(input_shape=(28, 28)),
    tf.keras.layers.Dense(128, activation="relu"),
    tf.keras.layers.Dense(10, activation="softmax")
])

model.summary()
```

## Step 8 — Compile the Model

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("Model compiled successfully!")
```

## Step 9 — Train the Model

```python
model.fit(
    X_train,
    y_train,
    epochs=5,
    validation_split=0.1
)
```

## Step 10 — Evaluate the Model

Now we test the model on images it did not see during training.

```python
loss, accuracy = model.evaluate(X_test, y_test)

print("Test Loss:", loss)
print("Test Accuracy:", accuracy)
```

## Step 11 — Make a Prediction

Let's ask our trained model to recognize one test image.

```python
prediction = model.predict(X_test[0:1])

predicted_digit = tf.argmax(prediction[0]).numpy()

print("Predicted digit:", predicted_digit)
print("Actual digit:", y_test[0])
```

```python
plt.imshow(X_test[0], cmap="gray")
plt.title(f"Predicted: {predicted_digit}")
plt.axis("off")
plt.show()
```
