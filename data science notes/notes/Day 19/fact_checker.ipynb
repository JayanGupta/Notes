```python
import os
from google.colab import userdata

os.environ["GROQ_API_KEY"] = userdata.get("GROQ_API_KEY")

os.environ["TAVILY_API_KEY"] = userdata.get("TAVILY_API_KEY")
```

```python
# Tool

from langchain_community.tools.tavily_search import TavilySearchResults

search = TavilySearchResults(max_results = 3)

tools = [search]
```

```python
# LLM

from langchain_groq import ChatGroq

llm = ChatGroq(model = "llama-3.1-8b-instant")
```

```python
from langchain.agents import create_agent
from langchain_core.messages import HumanMessage , SystemMessage

system_prompt = """
You are an jouralist who is expert in investigation and fact checking.
You are given a claim , you must use your search tool to find recent
and reliable sources. Your final response should be like :

Verdict : [TRUE , FALSE,  MISLEADING]

FACTS : [2-3 SENTENCES  explaning the truth]

SOURCES: [LIST OF URLs you found]
"""
```

```python
agent  = create_agent(llm , tools)
```

```python
claim = "Is is ethanol compulsory in India ?"
```

```python
response = agent.invoke({"messages": [SystemMessage(content = system_prompt ),
                           HumanMessage(content = claim)]})
```

```python
print(response["messages"][-1].content)
```
